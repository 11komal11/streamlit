```python
#Implementation of calculator using Main Function
''' use a main() function in Python to structure the program's entry point.'''
#choose an arithmetic operation (addition, subtraction, product or division
def add(x,y):
    return x+y

def subtract(x,y):
    return x-y

def multiply(x,y):
    return x*y
def divide(x,y):
    if y== 0:
        raise ValueError("Cannot divide by zero.")
    return x/y

#input two numbers to perform the operation
def main():
    print("Simple Calculator")
    print('''Select operation:
    1. Add
    2 Subtract
    3 Multiply
    4 Divide
    ''')
    #Get user Choice
    choice = input("Enter choice(1/2/3/4):")
    #get user input for numbers
    n1 = float(input("Enter first number:"))
    n2 = float(input("Enter second number:"))
    #perform the chosen operation
    if choice == '1':
        print(f"{n1}+{n2} = {add(n1,n2)}")
    elif choice == '2':
        print(f"{n1}-{n2} = {subtract(n1,n2)}")
    elif choice == '3':
        print(f"{n1}*{n2} = {multiply(n1,n2)}")
    if choice == '4':
        print(f"{n1}/{n2} = {divide(n1,n2)}")
    else:
        print("Invalid input! Please select a valid operation.")

if __name__ == "__main__":
    main()
```

```python
#lab44 Implementation of Python Exception Handling with calculating the area
''' how to handle logical errors (such as incorrect input values)
to manage exceptional cases or invalid input such as non-numeric values or negative numbers
Raise a ValueError when negative or invalid values are provided for dimensions.
The goal is to ensure that the program does not crash when encountering invalid input 
and provides appropriate error messages to the user
'''
#To calculate the area of a circle, rectangle and triangle
import math
def area_circle(radius):
    if radius <0:
        raise ValueError("Radius cannot be negative")
    return math.pi* radius**2

def area_rect(length, width):
    if length<0 or width<0:
        raise ValueError("Length and width cannot be negative")
    return length *width

def area_tri(base, height):
    if base<0 or height<0:
        raise ValueError("Base and height cannot be negative")
    return (base *height)/2

#Main function to handle user input and calculate area
def main():
    try:
        print('''Select the shape to calculate the area:
        1. Circle
        2. Rectangle
        3. Triangle
        ''')
        choice = int(input("Enter your choice (1/2/3):"))
        if choice == 1:
            radius = float(input("Enter the radius of the circle:"))
            area = area_circle(radius)
            print(f'The area of the circle is : {area:.2f}')
        elif choice == 2:
            length = float(input("Enter the length of the rectangle:"))
            width = float(input("Enter the width of th rectangle:"))
            area = area_rect(length, width)
            print(f'The area of the rectangle is : {area:.2f}')
        elif choice == 3:
            base = float(input("Enter the base of the triangle:"))
            height = float(input("Enter the height of the triangle:"))
            area = area_tri(base, height)
            print(f'The area of the triangle is : {area:.2f}')
        else:
            print("Invalid choice! Please select a valid option.")
    except ValueError as e:
        print(f"Error: {e}")
    except Exception as e:
        print(f"An unexpected error occurred:{e}")

#Execute the main function
if __name__ == "__main__":
    main()
```

```python
#lab 46 Implementation of Built-in Exceptions for various Error Conditions
import os 
def divide_numbers(x,y):
    try:
        return x/y
    except ZeroDivisionError:
        return "Error: Division by zero is not allowed."
def access_list_element(lst, index):
    try:
        return lst[index]
    except IndexError:
        return "Error:Index is out of range."
def get_dict_value(dct, key):
    try:
        return dct[key]
    except KeyError:
        return "Error: Key not found in dictionary"
def convert_to_integer(value):
    try:
        return int(value)
    except ValueError:
        return "Error: Cannot convert value to integer"
    except TypeError:
        return "Error: Invalid type for conversion"

#Demonstration of different exceptions
print(divide_numbers(10,0))     #ZeroDivisionError

my_list = [1,2,3]
print(access_list_element(my_list, 5)) #IndexError

my_dict = {'a':1, 'b': 2}
print(get_dict_value(my_dict, 'c'))   #KeyError

print(convert_to_integer('abc'))  #ValueError
print(convert_to_integer(3+4j))    #TypeError
        




        
```
